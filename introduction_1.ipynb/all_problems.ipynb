```python
#Count Vowels
word = "education" #the word we want to check.
vowels = "aeiouAEIOU" #list of all vowels (both small and capital).
vowel_count = 0 #   starting counters at zero.
consonants_count = 0 # starting counters at zero.
for char in word: #goes letter by letter: e, d, u, c, a, t, i, o, n.
    if char in vowels:#  checks if the letter is a vowel.
        vowel_count += 1#  If yes → increase vowel_count. → increase consonant_count.
    else:
        consonants_count += 1
print("vowels:",vowel_count)  #  Finally, print both counts.
print("cosonants:",consonants_count)
```

```python
# Palindrome Check
word = input("Enter a word: ")
# Reverse the word
reversed_word = word[::-1]  #reverses the string. Example: "madam" → "madam", "hello" → "olleh".
# Compare original and reversed
if word == reversed_word:
    print("Palindrome")
else:
    print("Not palindrome")
```

```python
#Student Marks Calculator ⭐ Ask for 5 subject marks.
# Student Marks Calculator (Beginner Level)

# Step 1: Ask for 5 subject marks
mark1 = int(input("Enter mark for subject 1: "))
mark2 = int(input("Enter mark for subject 2: "))
mark3 = int(input("Enter mark for subject 3: "))
mark4 = int(input("Enter mark for subject 4: "))
mark5 = int(input("Enter mark for subject 5: "))

# Step 2: Calculate total
total = mark1 + mark2 + mark3 + mark4 + mark5

# Step 3: Calculate average
average = total / 5

# Step 4: Find highest and lowest
highest = max(mark1, mark2, mark3, mark4, mark5)
lowest = min(mark1, mark2, mark3, mark4, mark5)

# Step 5: Pass/Fail (assuming pass if all marks >= 35)
if mark1 >= 35 and mark2 >= 35 and mark3 >= 35 and mark4 >= 35 and mark5 >= 35:
    result = "Pass"
else:
    result = "Fail"

# Step 6: Print results
print("Total Marks:", total)
print("Average Marks:", average)
print("Highest Mark:", highest)
print("Lowest Mark:", lowest)
print("Result:", result)
```

```python
# Simple Calculator 🎯

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: Division by zero is not allowed!"

# Ask user for operation
print("Choose operation:")
print("1. Add")
print("2. Subtract")
print("3. Multiply")
print("4. Divide")

choice = input("Enter choice (1/2/3/4): ")

# Ask for numbers
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Perform calculation
if choice == "1":
    print("Result:", add(num1, num2))
elif choice == "2":
    print("Result:", subtract(num1, num2))
elif choice == "3":
    print("Result:", multiply(num1, num2))
elif choice == "4":
    print("Result:", divide(num1, num2))
else:
    print("Invalid choice!")
```
