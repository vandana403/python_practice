```python
#Create a nested list containing 3 students and their marks.
students = [
    ["Vandana", 80],
    ["Rahul", 75],
    ["Priya", 90]
]

print(students)
```

```python
#Print the second student's name.
students = [
    ["Vandana", 80],
    ["Rahul", 75],
    ["Priya", 90]
]

print(students[1][0])
```

```python
#Print the third student's marks.
students = [
    ["Vandana", 80],
    ["Rahul", 75],
    ["Priya", 90]
]
print(students[2][1])
```

```python
students = [
    ["Vandana", 80],
    ["Rahul", 75],
    ["Priya", 90]
]
for student in students:
    print("name:",student[0])
```

```python
# List of student names
students = ["Alice", "Ben", "Carlos", "Dina"]

# Loop through the list and print each name
for name in students:
    print(name)
```

```python
student = {
    "name": "Vandana",
    "subjects": ["Python", "DBMS", "Java"],
    "marks": [80, 75, 90]
}
for key, value in student.items():
    print(f"{key.capitalize()}:{value}")
```
