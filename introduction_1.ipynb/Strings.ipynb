```python
#Strings
name = "vandana"
print(name)
```

```python
#Print the length of your name.
name = "vandana"
print(len(name))
```

```python
#Convert your name to uppercase.
name = "vandana"
print(name.upper())
```

```python
#Convert your name to lowercase.
name = "vandana"
print(name.lower())
```

```python
#Print the first character of your name.
name = "Navya"
print(name[0])
```

```python
#Full Name Ask the user for: First name Last name. Combine them and print the full name.
first_name = input("enter your first_name:")
second_name = input("enter your second_name:")
full_name = first_name + second_name
print(full_name)
```

```python
#Count Characters.Ask the user for a word and print its length.
word = input("enter one word:")
print(len(word))
```

```python
#Search.Ask the user for a sentence and a word.Check whether the word exists in the sentence.
sentence = input("enter your sentence:")
print("python" in sentence)
```

```python
#Reverse Thinking
word = "Python"
print(word[::-1])
#This method utilizes Python's built-in slice notation [start:stop:step].
#By leaving start and stop blank and setting the step to -1, you tell Python to read the string from right to left. 
```

```python
#Username.Ask the user for their name.
#Create a simple username by:
#converting the name to lowercase
#removing unnecessary spaces
name = input("enter your name:")
print(name.lower())
print(name.strip())
```

```python
name = input("Enter your name: ")
user_name = name.strip().lower()
print(user_name)
#.strip()  → remove extra outside spaces
#.lower()  → convert to lowercase
```
