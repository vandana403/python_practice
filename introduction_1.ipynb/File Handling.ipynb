```python
def divide_numbers():
    try:
        # Prompt the user for two numbers
        numerator = float(input("Enter the first number (numerator): "))
        denominator = float(input("Enter the second number (denominator): "))
        
        # Perform the division
        result = numerator / denominator
        
    except ValueError:
        # Handles cases where the user inputs text instead of numbers
        print("Error: Invalid input. Please enter valid numeric values.")
        
    except ZeroDivisionError:
        # Handles cases where the denominator is zero
        print("Error: Division by zero is not allowed.")
        
    else:
        # Executes if no exceptions were raised
        print(f"The result of the division is: {result}")

# Run the program
divide_numbers()
```
