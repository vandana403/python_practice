```python
#Multiplication Table
number = int(input("enter  a number:",))
print(f"Multiplication Table for {number}:")
for i in range(1,11):
    print(f"{number} x {i} = {number * i}")
```

```python
#Sum of Numbers from 1 to 5 --->with loop
count = 1
total = 0

while count <= 5:
    total = total + count 
    #print("Sum =", total)
    count = count + 1

print("Sum =", total)
```

```python
#Sum from 1 to N
n = int(input("enter n:",))
count = 1
total = 0
while count <= n:
    total = total + count
    count = count + 1
print("sum =",total)
```

```python
#Even/Odd Counter. Ask the user for N.
n = int(input("enter n:",))
even_count = 0
odd_count = 0 
for i in range(1,50):
    if i % 2 == 0:
        even_count = even_count + 1
    else:
        odd_count = odd_count + 1
print(even_count)
print(odd_count)
```
