```python
#Sum of Numbers from 1 to 5 --->with loop
count = 1
total = 0

while count <= 5:
    total = total + count 
    #print("Sum =", total)
    count = count + 1

print("Sum =", total)
```

```python
#Initialize variables
total_sum = 0
number = 1

# Execute the while loop
while number <= 5:
    total_sum += number  # Add current number to the total
    number += 1          # Increment the number by 1

print("The sum is:", total_sum)
```

```python
#break statement
# #Print numbers from 1 to 20, but stop when you reach 12.
count = 1
while count <= 20 :
    print(count)

    if count == 12:
        break
    count = count + 1
```

```python
#Print numbers from 10 to 1, but stop when the number becomes 6.
count = 10
while count >= 1 :
    print(count)

    if count == 6:
        break
    count = count - 1
```

```python
#continue Statement
#Print numbers from 1 to 20, but skip 10.
count = 0
while count < 20:
    count = count + 1

    if count == 10:  ########## here 10 is skipped
        continue
    print(count)
```

```python
#Print numbers from 1 to 15, but skip all even numbers.
count = 1
while count <= 15:
    if count % 2 == 0:
        count = count + 1 
        continue
    print(count)
    count = count + 1
```

```python
#Print numbers from 1 to 20, but skip numbers divisible by 3.
count = 1
while count <= 20 :
    if count % 3 == 0:
        count = count + 1
        continue
    print(count)
    count = count + 1
```

```python
#Count Digits in a Number
#Take a number from the user and count how many digits it has.
number = int(input("Enter a number: "))
count = 0

while count < 0 :
    count = count + 1
    number = number + count

print("Digits =", count)
```
